```python
import pandas as pd

print(pd.__version__)
print("Pandas working!")
```

```python
from sklearn.preprocessing import LabelEncoder

print("Scikit-learn working!")
```

```python
df = pd.read_csv("D:\sales_data.csv.csv")

df["Date"] = pd.to_datetime(df["Date"])
```

```python
import os

print(os.path.exists("models/random_forest_model.pkl"))
```

```python
import pickle

with open("models/random_forest_model.pkl", "rb") as f:
    model = pickle.load(f)

print("✅ Model loaded successfully!")
```

```python
df["Year"] = df["Date"].dt.year
df["Month"] = df["Date"].dt.month
df["Day"] = df["Date"].dt.day
df["DayOfWeek"] = df["Date"].dt.dayofweek
```

```python
categorical_columns = [
    "Store ID",
    "Product ID",
    "Category",
    "Region",
    "Weather Condition",
    "Seasonality"
]

for col in categorical_columns:
    le = LabelEncoder()
    df[col] = le.fit_transform(df[col])
```

```python
X = df[
    [
        "Store ID",
        "Product ID",
        "Category",
        "Region",
        "Inventory Level",
        "Units Sold",
        "Units Ordered",
        "Price",
        "Discount",
        "Weather Condition",
        "Promotion",
        "Competitor Pricing",
        "Seasonality",
        "Epidemic",
        "Year",
        "Month",
        "Day",
        "DayOfWeek",
    ]
]
```

```python
df["Predicted Demand"] = model.predict(X)
```

```python
def risk_level(row):
    if row["Inventory Level"] < row["Predicted Demand"]:
        return "Stockout Risk"
    elif row["Inventory Level"] > row["Predicted Demand"] * 1.5:
        return "Overstock Risk"
    else:
        return "Healthy"

df["Risk"] = df.apply(risk_level, axis=1)
```

```python
def recommendation(risk):
    if risk == "Stockout Risk":
        return "Reorder Stock"
    elif risk == "Overstock Risk":
        return "Offer Discount / Reduce Orders"
    else:
        return "No Action Required"

df["Recommendation"] = df["Risk"].apply(recommendation)
```
