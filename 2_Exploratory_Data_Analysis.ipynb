```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("D:\sales_data.csv.csv")

# Convert Date back to datetime
df["Date"] = pd.to_datetime(df["Date"])
```

```python
daily_demand = df.groupby("Date")["Demand"].sum()

plt.figure(figsize=(12,5))
plt.plot(daily_demand.index, daily_demand.values)
plt.title("Daily Demand Over Time")
plt.xlabel("Date")
plt.ylabel("Total Demand")
plt.xticks(rotation=45)
plt.show()
```

```python
category_demand = df.groupby("Category")["Demand"].sum()

plt.figure(figsize=(10,5))
category_demand.sort_values(ascending=False).plot(kind="bar")
plt.title("Demand by Category")
plt.xlabel("Category")
plt.ylabel("Total Demand")
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.hist(df["Inventory Level"], bins=20)
plt.title("Inventory Level Distribution")
plt.xlabel("Inventory Level")
plt.ylabel("Frequency")
plt.show()
```

```python
season_demand = df.groupby("Seasonality")["Demand"].sum()

plt.figure(figsize=(8,5))
season_demand.plot(kind="bar")
plt.title("Demand by Seasonality")
plt.xlabel("Season")
plt.ylabel("Total Demand")
plt.show()
```

```python
promotion = df.groupby("Promotion")["Demand"].mean()

plt.figure(figsize=(6,5))
promotion.plot(kind="bar")
plt.title("Promotion vs Average Demand")
plt.xlabel("Promotion")
plt.ylabel("Average Demand")
plt.show()
```

```python
inventory = df.groupby("Category")["Inventory Level"].mean()

plt.figure(figsize=(10,5))
inventory.plot(kind="bar")
plt.title("Average Inventory by Category")
plt.xlabel("Category")
plt.ylabel("Inventory Level")
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.scatter(df["Price"], df["Demand"], alpha=0.5)
plt.title("Price vs Demand")
plt.xlabel("Price")
plt.ylabel("Demand")
plt.show()
```

```python
import matplotlib.pyplot as plt

numeric = df.select_dtypes(include=["number"])

corr = numeric.corr()

plt.figure(figsize=(10,8))
plt.imshow(corr, cmap="coolwarm", aspect="auto")
plt.colorbar()

plt.xticks(range(len(corr.columns)), corr.columns, rotation=90)
plt.yticks(range(len(corr.columns)), corr.columns)

plt.title("Correlation Heatmap")
plt.show()
```
